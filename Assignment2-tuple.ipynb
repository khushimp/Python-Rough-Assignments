```python
#Assignment2

#create a tuple for courses
courses = ("Python", "Data Science", "Cloud Computing", "Cyber Security", "AI")
#Display the second course
print("Second course:", courses[1])
# Display the last two courses of tuple
print("Last two courses:", courses[-2:])
#Check whether Python is present in tuple
if "Python" in courses:
    print("Python is present in the course list.")
else:
    print("Python is not present in the course list.")
#Find the index of Cloud Computing in the tuple
print("Index of Cloud Computing in tuple:", courses.index("Cloud Computing"))
#Find total number of courses
print("Total number of courses:", len(courses))
#create a new tuple with old and new courses in it
courses+= ("Machine Learning", "Web Development")
#Create a new tuple without Cyber Security by using a list
#courses = tuple(course for course in courses if course != "Cyber Security")
y = list(courses)
y.remove("Cyber Security")
courses = tuple(y)
#Arrange courses alphabetically by interconverting back from a list
sorted_courses = tuple(sorted(courses))
#Create a backup copy of courses tuple
backup_courses = sorted_courses
#create tuple for Friend's courses
friend_courses = ("AI", "Blockchain", "Python")
#Combine student's courses and friend's courses into a new tuple 
combined_courses = sorted_courses + friend_courses
# Display final results
print("\nFinal course tuple:", courses)
print("Sorted tuple:", sorted_courses)
print("Backup tuple:", backup_courses)
print("Combined tuple:", combined_courses)
```
