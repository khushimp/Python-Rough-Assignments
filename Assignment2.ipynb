```python
#Assignment2

# Taking three numbers as input from the user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))
#Check equality conditions and display appropriate messages
if num1 == num2 == num3:
    print("All three numbers are equal.")
elif num1 == num2:
    print("The first and second numbers are equal.")
elif num2 == num3:
    print("The second and third numbers are equal.")
elif num1 == num3:
    print("The first and third numbers are equal.")
#Use if-else logic to find and display the largest number of 3 no.s
if num1 >= num2 and num1 >= num3:
    largest = num1
elif num2 >= num1 and num2 >= num3:
    largest = num2
else:
    largest = num3
print(f"The largest number among the three is: {largest}")
```
