```python
#Assignment3 

#define funtion to check if right angled triangle or not
def right_triangle(side1, side2, side3):
    #sort sides length-wise to find hypotenuse
    sides = sorted([side1, side2, side3])
    # to Check the Pythagorean theorem: a^2 + b^2 = c^2
    if sides[0]**2 + sides[1]**2 == sides[2]**2:
        return True
    else:
        return False
        
#enter sides of triangle
a = float(input("Enter the length of the first side: "))
b = float(input("Enter the length of the second side: "))
c = float(input("Enter the length of the third side: "))
# Call the function and print the result
if right_triangle(a,b,c)==True:
    print("The triangle is a right-angled triangle")
else:
    print("The triangle is not a right-angled triangle")
```
