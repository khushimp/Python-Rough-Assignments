```python
#Assignment 7

# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV dataset
df = pd.read_csv("D:/KHUSHI/MIT WPU/Python 1st yr/Data_Visualization/sample_data_assignment_07.csv")

# Display the dataset
print("Dataset:")
print(df)

# 1. Line Plot
plt.figure(figsize=(7, 4))
plt.plot(df["Student"], df["Python"], marker="o")

plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()

# 2. Bar Plot
plt.figure(figsize=(7, 4))
plt.bar(df["Student"], df["Maths"])

plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.show()

# 3. Scatter Plot
plt.figure(figsize=(7, 4))
plt.scatter(df["Maths"], df["Python"], s=70)

plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()

# 4. Calculate Correlation Matrix
correlation = df[["Maths", "Python", "Attendance"]].corr()

print("Correlation Matrix:")
print(correlation)

# 5. Heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(correlation, annot=True, cmap="Blues",
            vmin=-1, vmax=1)

plt.title("Correlation Heatmap")
plt.show()
```
